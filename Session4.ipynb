```python
import matplotlib.pyplot as plt

orders = [12, 15, 17, 14, 13, 16, 200, 18, 14, 15]

plt.figure(figsize=(8, 4))
plt.hist(orders, bins=8, edgecolor="black")
plt.title("Daily Order Counts")
plt.xlabel("Number of orders")
plt.ylabel("Number of days")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Answer:** The data is **right-skewed (positively skewed)**. Most values are between 12 and 18, but the unusually high value 200 stretches the distribution far to the right.

```python
followers = [5, 7, 8, 8, 9, 10, 12, 15, 95]

mean_followers = sum(followers) / len(followers)
sorted_followers = sorted(followers)
median_followers = sorted_followers[len(sorted_followers) // 2]

print("Mean:", round(mean_followers, 2), "thousand followers")
print("Median:", median_followers, "thousand followers")
```

**Answer:**
- Mean = **18.78 thousand**
- Median = **9 thousand**
- The mean is higher than the median.

This suggests a **right-skewed distribution**. The influencer with 95 thousand followers pulls the mean upward, while the median stays at the middle value.

```python
video_views = [1, 2, 2, 3, 4, 5, 6, 8, 12, 250]

plt.figure(figsize=(8, 4))
plt.hist(video_views, bins=8, edgecolor="black")
plt.title("Example YouTube Video Views")
plt.xlabel("Views (thousands)")
plt.ylabel("Number of videos")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Answer:** This is an illustrative example of **right-skewed data**. Most videos have low view counts, but one video has a very high view count. That unusually large value creates a long tail on the right. (These numbers are an example dataset, not actual YouTube account data.)

```python
ratings = [3, 3, 4, 4, 4, 5, 5, 5, 5, 5]

rating_values = [3, 4, 5]
rating_counts = [ratings.count(value) for value in rating_values]

plt.figure(figsize=(7, 4))
plt.bar(rating_values, rating_counts, edgecolor="black")
plt.xticks(rating_values)
plt.title("Frequency of Flipkart Product Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of products")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

**Answer:** The distribution is **left-skewed (negatively skewed)**. Most products have high ratings (especially 5), while fewer lower ratings extend toward the left.
