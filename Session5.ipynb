
**Population** means the complete group we want to study.  
**Sample** means a smaller part selected from that population.

### Example 1: Instagram followers
- **Population:** All followers of a particular Instagram account.
- **Sample:** 100 followers selected from that account to answer a survey.

### Example 2: Zomato restaurants
- **Population:** All restaurants listed on Zomato in Ahmedabad.
- **Sample:** 50 restaurants selected from the list to study their ratings.

### Example 3: Flipkart products
- **Population:** All mobile phones listed on Flipkart.
- **Sample:** 100 mobile phones selected to compare their prices and ratings.

```python
import random

# Create a list of 1,000 fake user IDs
user_ids = [f"user{i}" for i in range(1, 1001)]

# Select 50 random users
sample_users = random.sample(user_ids, 50)

# Print the selected users
print("Selected 50 users:")
for user in sample_users:
    print(user)
```

## 3. Why Use Sampling for Swiggy Delivery Times?

Swiggy receives a very large number of orders every day. Collecting and analyzing the delivery time of every order would take more time, storage, and effort. By selecting a sample of orders, we can estimate the average delivery time more quickly and with fewer resources.

## 4. Sampling Bias in the Spotify Survey

This introduces **selection bias**.

The survey is sent only to users who recently streamed Bollywood songs. These users are more likely to prefer Bollywood music, so other listeners—such as people who mainly listen to English, Punjabi, or classical music—are left out. As a result, the sample does not represent all Spotify users, and the survey results may be misleading.

## 5. Two Sampling Methods for BookMyShow

### Method 1: Random Sampling
We can make a list of all BookMyShow users and randomly select 500 users. Every user has an equal chance of being selected. We can then ask these users about their movie booking habits.

### Method 2: Stratified Sampling
First, divide BookMyShow users into groups based on age, such as 18–25, 26–40, and above 40. Then randomly select some users from each age group. This helps us include different age groups and compare their movie booking habits.
