# Variance & Standard Deviation Assignment

**Formulas used** (population version, since we have the complete data for each question):

- Mean: μ = Σx / N
- Variance: σ² = Σ(x − μ)² / N
- Standard deviation: σ = √σ²

*(If the data is treated as a sample, divide by N − 1 instead of N. Sample values are shown as a side note.)*

## Q1. Variance of weekly step counts (by hand)

Data: `[4500, 7000, 5000, 8000, 4000, 9000, 6000]`, N = 7

**Step 1: Mean**
Σx = 4500 + 7000 + 5000 + 8000 + 4000 + 9000 + 6000 = 43,500
μ = 43,500 / 7 = **6214.29**

**Step 2: Deviation from mean and squared deviation**

| Day | x | x − μ | (x − μ)² |
|---|---|---|---|
| 1 | 4500 | −1714.29 | 2,938,775.51 |
| 2 | 7000 | 785.71 | 617,346.94 |
| 3 | 5000 | −1214.29 | 1,474,489.80 |
| 4 | 8000 | 1785.71 | 3,188,775.51 |
| 5 | 4000 | −2214.29 | 4,903,061.22 |
| 6 | 9000 | 2785.71 | 7,760,204.08 |
| 7 | 6000 | −214.29 | 45,918.37 |

**Step 3: Sum of squared deviations**
Σ(x − μ)² = **20,928,571.43**

**Step 4: Variance**
σ² = 20,928,571.43 / 7 = **2,989,795.92**

*(Sample variance, dividing by 6, would be 3,488,095.24.)*

Check the same steps in code:

```python
steps = [4500, 7000, 5000, 8000, 4000, 9000, 6000]
n = len(steps)

mean = sum(steps) / n
print(f"Mean = {mean:.2f}")

total = 0
for x in steps:
    sq = (x - mean) ** 2
    total += sq
    print(f"{x}: ({x} - {mean:.2f})^2 = {sq:,.2f}")

print(f"Sum of squares = {total:,.2f}")
print(f"Variance = {total / n:,.2f}")
```

## Q2. Function: `calculate_standard_deviation(scores)`

```python
import math

def calculate_standard_deviation(scores):
    mean = sum(scores) / len(scores)
    variance = sum((x - mean) ** 2 for x in scores) / len(scores)
    return round(math.sqrt(variance), 2)

# Test cases
print(calculate_standard_deviation([50, 50, 50, 50]))       # 0.0
print(calculate_standard_deviation([2, 4, 4, 4, 5, 5, 7, 9]))  # 2.0
print(calculate_standard_deviation([70, 80, 90]))            # 8.16
print(calculate_standard_deviation([4500, 7000, 5000, 8000, 4000, 9000, 6000]))  # 1729.1
```

## Q3. Zomato spending: Friend A vs Friend B

```python
spend_a = [200, 200, 200, 200, 200]
spend_b = [100, 300, 150, 400, 50]

print("Friend A std dev:", calculate_standard_deviation(spend_a))
print("Friend B std dev:", calculate_standard_deviation(spend_b))
```

**Calculation for Friend B:** mean = 1000 / 5 = 200. Deviations: −100, 100, −50, 200, −150. Squares: 10000, 10000, 2500, 40000, 22500, which sum to 85,000. Variance = 85,000 / 5 = 17,000, so σ = √17,000 ≈ **130.38**.
**Friend A:** every value equals the mean (200), so all deviations are 0 and σ = **0**.

**Interpretation:** Both friends spend ₹200 per day on average, but Friend A is perfectly consistent (σ = 0) while Friend B's spending swings widely (σ ≈ ₹130) from ₹50 to ₹400. A lower standard deviation means values stay close to the mean, so Friend A has the more consistent spending.

## Q4. Startup salaries

```python
salaries = [28000, 29000, 27000, 60000, 26500, 27500]

n = len(salaries)
mean = sum(salaries) / n
variance = sum((x - mean) ** 2 for x in salaries) / n

print(f"Mean     = {mean:,.2f}")
print(f"Variance = {variance:,.2f}")
print(f"Std dev  = {variance ** 0.5:,.2f}")
```

**Result:** mean = ₹33,000, variance ≈ 146,416,667, standard deviation ≈ **₹12,100**.
*(Sample version: variance = 175,700,000, std dev ≈ ₹13,255.)*

**Interpretation:** The spread is high, and it is caused almost entirely by one salary (₹60,000). The other five employees earn between ₹26,500 and ₹29,000, so the mean (₹33,000) is pulled up and does not represent a typical employee. Without the ₹60,000 salary the std dev drops to about ₹860, which shows the team is fairly uniform except for one outlier, likely a founder or senior hire.
